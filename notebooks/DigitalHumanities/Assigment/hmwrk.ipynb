```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("hmwrk.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
decade_counts = df['Decade'].value_counts().sort_index()

decade_counts.plot(kind='bar')

plt.title('Number of Events by Decade')
plt.xlabel('Decade')
plt.ylabel('Number of Events')

plt.show()
```

# This chart shows how the number of recorded events changes across different decades. Some decades contain more events than others, which may indicate periods of greater political or military activity in the dataset. The distribution is not even, so certain decades are represented much more strongly.

```python
type_counts = df['Type'].value_counts()

type_counts.plot(kind='bar')

plt.title('Number of Events by Type')
plt.xlabel('Event Type')
plt.ylabel('Number of Events')

plt.show()
```

# This chart compares the different types of historical events in the dataset. Some types appear more frequently than others. This suggests that the dataset focuses more strongly on certain categories of events.

```python
period_counts = df['Period'].value_counts()

period_counts.plot(kind='pie', autopct='%1.1f%%')

plt.title('Events by Historical Period')
plt.ylabel('')

plt.show()
```

# The chart shows the proportion of events belonging to each historical period. Some periods make up a larger part of the dataset than others. This helps show which periods are most strongly represented in the data.

# Conclusion, The visualisations show that the dataset is unevenly distributed across decades, event types and historical periods. Some categories are represented much more strongly than others, which helps identify the main patterns in the dataset.
