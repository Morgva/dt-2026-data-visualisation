```python
import pandas as pd
df = pd.read_csv('Newthing.csv')
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
avg_practice = df.groupby('Student')['Hours_Studied'].mean().reset_index()
avg_practice.plot(kind='bar', x='Student', y='Hours_Studied', title='Average Hours Studied per Student')
plt.ylabel('Average Hours Studied')
```

```python
avg_practice = df.groupby('Student')['Hours_Studied'].mean().reset_index()
avg_practice = avg_practice.sort_values(by='Hours_Studied')

avg_practice.plot(kind='bar', x='Student', y='Hours_Studied', title='Average Hours Studied per Student')
plt.ylabel('Average Hours Studied')
```
