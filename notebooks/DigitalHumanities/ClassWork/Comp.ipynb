```python
import pandas as pd
df = pd.read_csv('Newthing.csv')
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
avg_practice = df.groupby('Student')['Hours_Studied'].mean().reset_index()
avg_practice.plot(kind='bar', x='Student', y='Hours_Studied', title='Average Hours Studied per Student')
plt.ylabel('Average Hours Studied')
```

```python
avg_practice = df.groupby('Student')['Hours_Studied'].mean().reset_index()
avg_practice = avg_practice.sort_values(by='Hours_Studied')

avg_practice.plot(kind='bar', x='Student', y='Hours_Studied', title='Average Hours Studied per Student', color='purple')
plt.ylabel('Average Hours Studied')
```

```python
avg_practice = df.groupby('Hours_Studied')['Grade'].mean().reset_index()
avg_practice = avg_practice.sort_values(by='Grade')
avg_practice.plot(kind='scatter', x='Hours_Studied', y='Grade', title='Average Grade per Student', color='purple', xlabel='Hours Studied')
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"])
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Hours Studied vs Grade")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot of Hours Studied and Grade")
plt.ylabel("Values")
plt.show()
```
