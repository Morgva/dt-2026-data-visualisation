```python
# can use to work with tabulat data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
df.describe() #Show some basic statistics
```

```python
import matplotlib.pyplot as plt #matplotlib helps to creates charts and visualizations
df.plot(kind="bar", x="Name", y="Years_of_practice", title="Years of practice by Name", legend=False, ylabel="Years of Practice")
plt.show()
```

```python
df.plot(kind="bar", x="Name", y="Age", title="Age by Name", legend=False, ylabel="Age")
plt.show()
```

```python
df.plot(kind="pie", y="Age", title="Age distribution", legend=False, labels=df["Name"], autopct='%1.1f%%')
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60)
plt.xlabel("Age")
plt.ylabel("Years of Practice")
plt.title("Age vs Years of Practice")
plt.show()
```

```python
import seaborn as sns
sns.boxplot(data=df, y="Age")
plt.title("Box plot of Age vs Years of Practice")
```

```python
plt.figure(figsize=(10, 6))
df[["Age", "Years_of_practice"]].boxplot()
plt.title("Box plot of Age and Years of Practice")
plt.show()
```

```python
avg_age = df.groupby("Sport")["Age"].mean()
plt.bar(avg_age.index, avg_age.values)
plt.title("Average Age by Sport")
plt.xlabel("Sport")
plt.ylabel("Average Age")
plt.xticks(rotation=45)
```

```python
avg_years_of_practice = df.groupby("Sport")["Years_of_practice"].mean()
plt.bar(avg_years_of_practice.index, avg_years_of_practice.values)
plt.title("Average Years of Practice by Sport")
plt.xlabel("Sport")
plt.ylabel("Average Years of Practice")
plt.xticks(rotation=45)
```
